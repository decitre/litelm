```python
from litelm import LLM
llm = await LLM.create(generator_model="gpt2")
```

```python
text = await llm.generate("Bindings are useful becasue")
print(text)
```

```python
vec = await llm.embed("Hello World")
print(f"vector dimension: {len(vec)}")
print(f"first 10 dimensions: {vec[:10]}")
```

```python
docs = [
    "JupyterLite runs entirely in the browser.",
    "Python is widely used for machine learning.",
    "TypeScript is great for frontend applications.",
    "Transformers are powerful models for NLP."
]
results = await llm.similarity_search(
    "Where does JupyterLite run?",
    docs
)

for doc, score in results:
    print(score, "->", doc)
```

```python
llm3 = await LLM.create(generator_model="distilgpt2")
text = await llm3.generate("Bindings are useful becasue")
print(text)
```

```python
async def rag(llm, query, docs):
    results = await llm.similarity_search(query, docs)
    top_docs = [doc for doc, _ in results[:2]]
    context = "\\n".join(top_docs)
    prompt = f"""
Context:
{context}

Question:
{query}

Answer:
"""
    return await llm.generate(prompt)
```

```python
answer = await rag(llm, "Where does JupyterLite run?", docs)
print(answer)
```

```python
answer = await rag(llm3, "Where does JupyterLite run?", docs)
print(answer)
```
